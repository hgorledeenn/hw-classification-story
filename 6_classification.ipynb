```python
import pandas as pd
from plotnine import *
```

```python
df = pd.read_csv('data/poets_full_dataset.csv')
```

```python
df.shape
```

```python
df = df[~df['Title'].str.contains('audio only')]
```

```python
df_2000 = df[df['Year']>=2000]
```

```python
df_2000.shape
```

```python
plot_2000 = (
    ggplot(df_2000) +
    aes(x='Year') + 
    geom_histogram(bins=26)
)
plot_2000
```

```python
df_sample = df_2000.sample(n=375, random_state=4910459, replace=False)
```

```python
plot_sample = (
    ggplot(df_sample) + 
    aes(x='Year') +
    geom_histogram(bins=26)
)
plot_sample
```

```python
df_sample.to_csv('data/df_sample.csv', index=False)
```
