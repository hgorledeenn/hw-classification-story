### Comparing my manual labelling of poems vs LLM labeling

```python
import pandas as pd
```

```python
df_human = pd.read_csv('data/df_sample.csv')
```

```python
df_human.head(1)
```

```python
df_llm = pd.read_csv('data/df_llm_labels_1.csv')
```

```python
df_llm['llm'] = df_llm['llm'].replace({True: 'immigration', False: 'not immigration'})
```

```python
df_llm.head(1)
```

```python
df_merged = pd.merge(df_human, df_llm, on=['Title', 'Poet', 'Year', 'URL', 'Poem Text'], how='inner')

df_merged_nona = df_merged[~df_merged['llm'].isna()]
```

```python
df_merged.head(3)
```

```python
# Define positive class as "Aggravated Assault"
false_positives = df_merged_nona[
    (df_merged_nona['llm'] == True)
    & (df_merged_nona['label'] != 'immigration')
]

false_negatives = df_merged_nona[
    (df_merged_nona['llm'] != True)
    & (df_merged_nona['label'] == 'immigration')
]

cols_to_show = [
    'Title',
    'label',
    'llm',
    'reason',
]

# Show full text in dataframe cells
pd.set_option('display.max_colwidth', None)
```

```python
# use sklearn to calculate precision, recall, f1 and accuracy
from sklearn.metrics import classification_report
print(classification_report(df_merged_nona['label'], df_merged_nona['llm']))
```

```python
print(f"False positives: {len(false_positives)}")
false_positives[cols_to_show]
```

```python
print(f"False negatives: {len(false_negatives)}")
false_negatives[cols_to_show]
```
